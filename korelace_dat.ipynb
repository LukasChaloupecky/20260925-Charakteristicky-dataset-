```python
import pandas as pd
from scipy import stats

df = pd.read_csv(
    "/content/data/Drowning&Cage.csv",
    encoding="utf-8-sig",
    index_col=0,
).T.astype(int)

df.columns = ["utopeni", "cage movies"]
df.index.name = "rok"

display(df.head())

r, p = stats.pearsonr(df["cage movies"], df["utopeni"])
print(f"r = {r:.3f}, R² = {r**2:.3f}, p = {p:.4f}")
```
