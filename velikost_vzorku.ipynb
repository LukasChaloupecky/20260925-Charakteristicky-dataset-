```python
import pandas as pd
import numpy as np
import math


# The dataset for the testing - JJust randomized
df = pd.DataFrame({
    "jmeno": [f"student_{i}" for i in range(1, 1201)],
    "rocnik": [1, 2, 3, 4] * 300
})

df.head()



z = 1.96       # 95% jistota
p = 0.5        # nejkonzervativnější odhad podílu spokojených
N = 1200       # počet studentů
navratnost = 0.6

def velikost_vzorku(e, z=z, p=p, N=N):
    n0 = z**2 * p * (1 - p) / e**2           # základní velikost (nekonečná populace)
    n = n0 / (1 + (n0 - 1) / N)              # korekce pro konečnou populaci
    return n0, n

for e in (0.05, 0.03):
    n0, n = velikost_vzorku(e)
    oslovit = math.ceil(math.ceil(n) / navratnost)
    print(f"±{e*100:.0f} %: n0 = {n0:.2f} -> {math.ceil(n0)}, "
          f"n = {n:.2f} -> {math.ceil(n)}, "
          f"oslovit při {navratnost:.0%} návratnosti: {oslovit}, "
          f"podíl n0 na populaci: {n0/N:.0%}")
```
