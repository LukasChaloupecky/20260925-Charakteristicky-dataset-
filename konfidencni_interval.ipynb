```python
import pandas as pd
import pandas as pd
import numpy as np
from scipy import stats
from google.colab import files


# df = pd.read_csv("Konfident_Interval_dataset.csv", sep="|", encoding="utf-8-sig", index_col=0).T.astype(float)
# df.columns = ["Měření", "Doba (min)"]

# df.head()

# 2) Načtení: data jsou v řádku, takže transponujeme
df = pd.read_csv("/content/data/Konfident_Interval_dataset.csv", encoding="utf-8-sig", index_col=0)
x = df.loc["Doba (min)"].astype(int).to_numpy()

df.head()

# 3) Průměr, směrodatná odchylka, standardní chyba
n = len(x)
mean = x.mean()
sd = x.std(ddof=1)          # ddof=1 = výběrová směrodatná odchylka (jako STDEV.S)
se = sd / np.sqrt(n)

# 4) 95% CI podle zadání (z = 1,96)
lo, hi = mean - 1.96 * se, mean + 1.96 * se

# (volitelně) přesnější varianta s t-rozdělením
t = stats.t.ppf(0.975, df=n - 1)
lo_t, hi_t = mean - t * se, mean + t * se

print(f"n = {n}")
print(f"průměr = {mean:.2f}, sd = {sd:.2f}, SE = {se:.3f}")
print(f"95% CI (z=1,96): ({lo:.2f}; {hi:.2f})")
print(f"95% CI (t):      ({lo_t:.2f}; {hi_t:.2f})")
print("90 min uvnitř intervalu?", lo <= 90 <= hi)
```
